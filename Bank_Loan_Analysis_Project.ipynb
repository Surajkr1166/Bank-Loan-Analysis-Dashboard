# ***IMPORT LIBRARIES***


```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px
import warnings
warnings.filterwarnings('ignore')
%matplotlib inline
sns.set_palette("husl")
print("All libraries loaded!")
```

**Dataset Upload & Load**

```python
from google.colab import files
uploaded = files.upload()


filename = list(uploaded.keys())[0]
df = pd.read_excel(filename)
print(f"✅ Dataset loaded: {df.shape[0]} rows, {df.shape[1]} columns")
print("\nFirst 5 rows:")
print(df.head())
```

**Data Cleaning**

```python
# Date columns fix
df['issue_date'] = pd.to_datetime(df['issue_date'], errors='coerce')
df['int_rate'] = df['int_rate'].astype(float)  # % remove
df['loan_status'] = df['loan_status'].astype('category')

# Good/Bad loans
good_loans = df[df['loan_status'].isin(['Fully Paid', 'Current'])]
bad_loans = df[df['loan_status'] == 'Charged Off']

print("✅ Data cleaned!")
print(f"Total loans: {len(df):,}")
print(f"Good loans: {len(good_loans):,} ({len(good_loans)/len(df)*100:.1f}%)")
print(f"Bad loans: {len(bad_loans):,} ({len(bad_loans)/len(df)*100:.1f}%)")
print("\nSample loan_status:")
print(df['loan_status'].value_counts())
```

**KPIs Table**

```python
latest_month = df['issue_date'].dt.to_period('M').max()
print(f"Latest month: {latest_month}")

# Calculate KPIs
kpis = {
    'Total Applications': len(df),
    'MTD Applications': len(df[df['issue_date'].dt.to_period('M') == latest_month]),
    'Total Funded ($)': df['loan_amount'].sum(),
    'MTD Funded ($)': df[df['issue_date'].dt.to_period('M') == latest_month]['loan_amount'].sum(),
    'Total Received ($)': df['total_payment'].sum(),
    'Avg Interest Rate (%)': df['int_rate'].mean(),
    'Good Loan %': len(good_loans)/len(df)*100
}

kpi_df = pd.DataFrame(list(kpis.items()), columns=['Metric', 'Value'])
print("## KEY PERFORMANCE INDICATORS")
kpi_df.style.format({'Value': '{:,.0f}'})
```

**Monthly Trend Chart**

```python
df_monthly = df.groupby(df['issue_date'].dt.to_period('M')).agg({
    'id': 'count',
    'loan_amount': 'sum',
    'total_payment': 'sum'
}).reset_index()
df_monthly.columns = ['Month', 'Apps', 'Funded', 'Received']
df_monthly['Month'] = df_monthly['Month'].astype(str)

fig = px.line(df_monthly, x='Month', y=['Apps', 'Funded', 'Received'],
              title=' Monthly Trends: Applications, Funded Amount, Received',
              labels={'value': 'Count / $'})
fig.update_layout(xaxis_tickangle=-45, height=500)
fig.show()
```

 **Top Charts (Regional + Purpose)**

```python
# Regional Top 10
state_top = df.groupby('address_state')['loan_amount'].sum().nlargest(10).reset_index()
fig1 = px.bar(state_top, x='address_state', y='loan_amount',
              title=' Top 10 States by Funded Amount')

# Purpose
purpose_top = df['purpose'].value_counts().head(8)
fig2 = px.pie(values=purpose_top.values, names=purpose_top.index,
              title=' Top Loan Purposes')

fig1.show()
fig2.show()
```

**Final - Download Report**

```python
# Save KPIs as CSV
kpi_df.to_csv('Bank_Loan_KPIs.csv', index=False)
files.download('Bank_Loan_KPIs.csv')

# Full summary
print("🎉 Project Complete! Resume mein add karo:")
print("- End-to-End Bank Loan Analysis")
print("- Python: Pandas, Plotly, Seaborn")
print("- KPIs + 5+ Visualizations")
```
